```python
# MINIMAL CODE: Only load data + beautiful sentiment scatter plot (OR vs CG)

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from textblob import TextBlob
import nltk

# Fix the NLTK error — download only what's needed
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)  # This is the missing one!

# Load your dataset
df = pd.read_csv('fake reviews dataset.csv')

# Quick feature: sentiment + length
df['sentiment_polarity'] = df['text_'].apply(lambda x: TextBlob(x).sentiment.polarity)
df['review_length'] = df['text_'].apply(len)

# Scatter plot — clean and beautiful
plt.figure(figsize=(12, 8))

# Original (OR) in green, Computer-Generated (CG) in red
plt.scatter(df[df['label'] == 'OR']['sentiment_polarity'],
            df[df['label'] == 'OR']['review_length'],
            color='#2ecc71', alpha=0.7, s=50, label='Original Reviews (OR)', edgecolor='black', linewidth=0.3)

plt.scatter(df[df['label'] == 'CG']['sentiment_polarity'],
            df[df['label'] == 'CG']['review_length'],
            color='#e74c3c', alpha=0.7, s=50, label='Fake / Generated Reviews (CG)', edgecolor='black', linewidth=0.3)

# Neutral line
plt.axvline(0, color='gray', linestyle='--', alpha=0.8)

# Styling
plt.title('Sentiment Polarity vs Review Length\nOriginal vs Computer-Generated Reviews', fontsize=18, pad=20)
plt.xlabel('Sentiment Polarity  (Negative ←  —  → Positive)', fontsize=14)
plt.ylabel('Review Length (characters)', fontsize=14)
plt.legend(fontsize=12, frameon=True, fancybox=True, shadow=True)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# PUNCTUATION CHECK – CG reviews love exclamation marks and perfect grammar
import string

def count_punctuation(text):
    return sum(1 for char in text if char in string.punctuation)

def count_exclamation(text):
    return text.count('!')

df['punct_count'] = df['text_'].apply(count_punctuation)
df['exclamation_count'] = df['text_'].apply(count_exclamation)
df['has_multiple_exclamations'] = df['text_'].str.contains(r'!!+', regex=True)

print("=== PUNCTUATION ABUSE CHECK ===")
print(df.groupby('label')[['punct_count', 'exclamation_count']].mean().round(2))

print("\nReviews with '!!!' or more (classic fake review sign):")
print(df[df['has_multiple_exclamations']]['label'].value_counts())
```
